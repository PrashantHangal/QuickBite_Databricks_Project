Loading all raw data files into Bronze layer

```python
df_customer = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/uc_quickbite/default/qbvolume/dim_customer.csv")
)

df_customer.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("uc_quickbite.bronze_ingestion.dim_customer")
```
